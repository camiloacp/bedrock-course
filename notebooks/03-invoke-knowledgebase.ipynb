```python
import boto3

session = boto3.session.Session( )
bedrock_agent_runtime_client = boto3.client('bedrock-agent-runtime')
```

```python
# 1. Definición de variables
query = "How we can used embeddings to create recommended systems?"
max_results = 3
kb_id = "ECSKINQEOJ"

# Asegúrate de usar un model_id válido en Amazon Bedrock
# Opción A: Modelo Amazon Nova (si está disponible en tu región)
model_id = "amazon.nova-lite-v1:0" 

# Opción B: Modelo Claude 3 Haiku (muy rápido y excelente para RAG)
#model_id = "anthropic.claude-3-haiku-20240307-v1:0"

model_arn = f'arn:aws:bedrock:us-east-1::foundation-model/{model_id}'

# Corrección: Corrección de un pequeño typo en "assistant" y "addressing"
default_prompt = """
You are a helpful assistant who helps users address questions related to AI. Only refer to the provided data
while responding to the questions.
$search_results$
"""

# 2. Llamada al cliente (asumiendo que bedrock_agent_runtime_client ya fue creado)
# Corrección: El método es retrieve_and_generate (sin la 't' extra al final)
response = bedrock_agent_runtime_client.retrieve_and_generate(
    input={
        'text': query
    },
    retrieveAndGenerateConfiguration={
        'type': 'KNOWLEDGE_BASE',
        'knowledgeBaseConfiguration': {
            'knowledgeBaseId': kb_id,
            'modelArn': model_arn, 
            'retrievalConfiguration': {
                'vectorSearchConfiguration': {
                    'numberOfResults': max_results
                }
            },
            'generationConfiguration': {
                'promptTemplate': {
                    'textPromptTemplate': default_prompt
                }
            }
        }
    }
)

# 3. Extracción del resultado
generated_text = response['output']['text']

print(generated_text)
```
